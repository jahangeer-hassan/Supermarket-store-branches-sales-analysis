## Store Area vs Store Sales Scatter Plot
## Is there any relationship between store area and sales?

```
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(
    df["Store_Area"],
    df["Store_Sales"],
    color="royalblue",
    edgecolor="black",
    alpha=0.7
)

plt.xlabel("Store Area", fontsize=12)
plt.ylabel("Store Sales", fontsize=12)
plt.title("Store Area vs Store Sales", fontsize=15)

plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

## Daily Customer Count vs Store Sales Scatter Plot
## Is there any relationship between daily customer count and sales?

```
plt.figure(figsize=(8,5))

plt.scatter(
    df["Daily_Customer_Count"],
    df["Store_Sales"],
    color="darkorange",
    edgecolor="black",
    alpha=0.7
)

plt.xlabel("Daily Customer Count", fontsize=12)
plt.ylabel("Store Sales", fontsize=12)
plt.title("Customer Count vs Store Sales", fontsize=15)

plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

## Items Available Distribution Histogram
## What is the distribution of items available?

```
plt.figure(figsize=(8,5))

plt.hist(
    df["Items_Available"],
    bins=15,
    color="mediumseagreen",
    edgecolor="black"
)

plt.xlabel("Items Available", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.title("Items Available Distribution", fontsize=15)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

## Store Sales Distribution Histogram
## What is the distribution of store sales?

```
plt.figure(figsize=(8,5))

plt.hist(
    df["Store_Sales"],
    bins=15,
    color="crimson",
    edgecolor="black"
)

plt.xlabel("Store Sales", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.title("Store Sales Distribution", fontsize=15)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

## Correlation Heatmap
## What patterns can be observed from the correlation heatmap?

```
import seaborn as sns

plt.figure(figsize=(8,5))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    linewidths=1,
    fmt=".2f"
)

plt.title("Correlation Heatmap", fontsize=15)

plt.show()
```
